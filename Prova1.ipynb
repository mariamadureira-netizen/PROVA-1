# parte 1
1 - A Fila (Queue - FIFO) é a mais adequada porque os chamados devem ser atendidos rigorosamente na ordem em que chegam, garantindo que o primeiro a entrar seja o primeiro a sair.

2 - A Pilha (Stack - LIFO) é a mais adequada porque o sistema precisa desfazer primeiro o atendimento mais recente, ou seja, o último elemento a entrar no histórico deve ser o primeiro a sair.

3 - Fila de espera: [103 - Carla - Baixa, 104 - Daniel - Alta]
Chamados já atendidos: [101 - Ana - Média, 102 - Bruno - Alta]

```python
# parte 2

class CentralAtendimento:
    def __init__(self):
        self.fila_espera = []
        self.chamados_atendidos = []

    def adicionar_chamado(self, codigo, cliente, prioridade):
        chamado = {"codigo": codigo, "cliente": cliente, "prioridade": prioridade}
        self.fila_espera.append(chamado)

    def atender_proximo(self):
        if len(self.fila_espera) > 0:
            chamado = self.fila_espera.pop(0)
            self.chamados_atendidos.append(chamado)
            return chamado
        return None

    def desfazer_ultimo_atendimento(self):
        if len(self.chamados_atendidos) > 0:
            chamado = self.chamados_atendidos.pop()
            self.fila_espera.insert(0, chamado)
            return chamado
        return None

    def buscar_chamado(self, codigo):
        for chamado in self.fila_espera:
            if chamado["codigo"] == codigo:
                return chamado
        return None

    def exibir_situacao(self):
        print("Aguardando:", [c["codigo"] for c in self.fila_espera])
        print("Atendidos:", [c["codigo"] for c in self.chamados_atendidos])
```

```python
class CentralAtendimentoOtimizada:
    def __init__(self):
        self.fila_espera = []
        self.chamados_atendidos = []
        # Dicionário auxiliar para busca em O(1)
        self.tabela_busca = {}

    def adicionar_chamado(self, codigo, cliente, prioridade):
        chamado = {"codigo": codigo, "cliente": cliente, "prioridade": prioridade}
        self.fila_espera.append(chamado)
        # Adiciona ao dicionário de busca
        self.tabela_busca[codigo] = chamado

    def atender_proximo(self):
        if len(self.fila_espera) > 0:
            chamado = self.fila_espera.pop(0)
            self.chamados_atendidos.append(chamado)
            # Remove do dicionário de busca, pois não está mais aguardando na fila
            self.tabela_busca.pop(chamado["codigo"], None)
            return chamado
        return None

    def desfazer_ultimo_atendimento(self):
        if len(self.chamados_atendidos) > 0:
            chamado = self.chamados_atendidos.pop()
            self.fila_espera.insert(0, chamado)
            # Coloca de volta no dicionário de busca
            self.tabela_busca[chamado["codigo"]] = chamado
            return chamado
        return None

    def buscar_chamado(self, codigo):
        # Busca direta no dicionário em O(1)
        return self.tabela_busca.get(codigo, None)

    def exibir_situacao(self):
        print("Aguardando:", [c["codigo"] for c in self.fila_espera])
        print("Atendidos:", [c["codigo"] for c in self.chamados_atendidos])
```

```python
# Demonstração do funcionamento da busca otimizada
central_opt = CentralAtendimentoOtimizada()

central_opt.adicionar_chamado(101, "Ana", "Média")
central_opt.adicionar_chamado(102, "Bruno", "Alta")
central_opt.adicionar_chamado(103, "Carla", "Baixa")
central_opt.adicionar_chamado(104, "Daniel", "Alta")

central_opt.atender_proximo()
central_opt.atender_proximo()
central_opt.desfazer_ultimo_atendimento()
central_opt.atender_proximo()

central_opt.exibir_situacao()

# A busca aqui ocorre de forma instantânea em O(1)
print("Resultado da busca rápida pelo 104:", central_opt.buscar_chamado(104))
```
